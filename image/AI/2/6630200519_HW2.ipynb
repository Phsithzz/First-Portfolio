# import BeautifulSoup และ requests เพื่อใช้ในการดึงข้อมูลจากหน้าเว็บ

```python
from bs4 import BeautifulSoup
import requests
```

# ทำการโหลดหน้าเว็บเพื่อเตรียมสำหรับ Scrapping

```python

url = 'https://www.bbc.com/news'
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
```

# ทำการ Scrap หัวข้อข่าวและประเภทของทวีปข่าวบันทึกลงใน list

```python
Data1 = soup.find_all("h2", class_="sc-8ea7699c-3 kwWByH")
Data2 = soup.find_all("span", class_="sc-6fba5bd4-2 bHkTZK")
Name = []
continent= []

for list in Data1:
    if list:
        Name.append(list.get_text())

for list in Data2:
    if list:
        continent.append(list.get_text())


list2 = [[a, b] for a, b in zip(Name, continent)]
print(list2)
```

# บันทึงลงในไฟล์ csv

```python
import csv

with open("AI_news.csv", "w", encoding = "UTF-8", newline = "\n") as file_csv:
  textcsv_writer = csv.writer(file_csv)
  textcsv_writer.writerow(["Headline", "continent"])
  textcsv_writer.writerows(list2)

```
